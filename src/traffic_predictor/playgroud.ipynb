```python
from dataset_reader import CsvReplayReader

CSV_FILE = "traffic_dataset.csv"

TIME_SCALE = 1.0
VERBOSE = True
reader = CsvReplayReader(CSV_FILE, time_scale=TIME_SCALE, verbose=VERBOSE)

while True:
    row = reader.getNext()
    if row is None:
        break

    if not row.should_transmit:
        continue

    # row.payload is exactly the payload string you were sending
    # row.relative_time is the scheduling offset if you want to re-use it
    #print(f"Row {row.row_index}: payload={row.payload}, rel_t={row.relative_time}")
    print(f"Row {row.row_index}: rel_t={row.relative_time}")
```
